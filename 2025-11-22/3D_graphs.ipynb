```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

def peaks(x, y):
    """
    Implementación de la función 'peaks' de MATLAB.
    Esta función genera una superficie con picos y valles,
    compuesta por una suma de funciones gaussianas y exponenciales.
    """
    z = 3 * (1 - x)**2 * np.exp(-(x**2) - (y + 1)**2) - \
        10 * (x/5 - x**3 - y**5) * np.exp(-x**2 - y**2) - \
        1/3 * np.exp(-(x + 1)**2 - y**2)
    return z

def generar_grafico_3d_identico():
    """
    Genera un gráfico de superficie 3D que reproduce exactamente
    la imagen proporcionada, utilizando la función 'peaks'.
    """
    # 1. Crear los datos de la malla
    # La imagen muestra la función 'peaks' en su rango típico (aprox. -3 a 3),
    # aunque los ejes están etiquetados de -1 a 1. Usamos el rango [-3, 3]
    # para obtener la forma correcta.
    # Se usa una resolución de 35x35 puntos para coincidir con la densidad de la cuadrícula de la imagen.
    x = np.linspace(-3, 3, 35)
    y = np.linspace(-3, 3, 35)
    X, Y = np.meshgrid(x, y)
    Z = peaks(X, Y)

    # 2. Crear la figura y los ejes 3D
    fig = plt.figure(figsize=(8, 6))
    ax = fig.add_subplot(111, projection='3d')

    # 3. Graficar la superficie
    # - cmap='cool': Mapa de colores de cian a magenta, similar al de la imagen.
    # - edgecolor='black': Color de las líneas de la cuadrícula sobre la superficie.
    # - linewidth=0.5: Grosor de las líneas de la cuadrícula.
    # - alpha=1.0: Opacidad de la superficie.
    # - rstride=1, cstride=1: Frecuencia de las líneas de la cuadrícula (cada fila/columna).
    surf = ax.plot_surface(X, Y, Z, cmap='cool', edgecolor='black', linewidth=0.5, alpha=1.0, rstride=1, cstride=1)

    # 4. Configurar los ejes para que coincidan visualmente con la imagen
    # Establecer los límites de los ejes para mostrar los datos completos.
    ax.set_xlim(-3, 3)
    ax.set_ylim(-3, 3)
    ax.set_zlim(-1, 1.2) # Un poco más de espacio en Z para las etiquetas

    # Ajustar manualmente las etiquetas de los ejes para que coincidan con la imagen (-1 a 1).
    # Esto se hace mapeando las posiciones de las marcas en el rango de datos [-3, 3]
    # a las etiquetas deseadas [-1, 1].
    ticks_pos = np.linspace(-3, 3, 5)
    tick_labels = ['-1', '-0.5', '0', '0.5', '1']
    
    ax.set_xticks(ticks_pos)
    ax.set_xticklabels(tick_labels)
    
    ax.set_yticks(ticks_pos)
    ax.set_yticklabels(tick_labels)
    
    # Las marcas en Z de la imagen son -0.5, 0, 0.5, 1.
    # También hay una marca inferior que parece ser -1.
    zticks_pos = np.array([-1, -0.5, 0, 0.5, 1])
    ax.set_zticks(zticks_pos)

    # 5. Ajustar la vista y el estilo
    # Establecer el ángulo de visión (elevación y azimut) para que coincida con la perspectiva de la imagen.
    ax.view_init(elev=30, azim=-45)

    # Configurar la cuadrícula de fondo de los ejes.
    ax.grid(True, linestyle=':', linewidth=0.5, color='gray')

    # Hacer que los paneles de fondo de los ejes sean transparentes para un estilo más limpio,
    # similar al gráfico de MATLAB original.
    ax.xaxis.set_pane_color((1.0, 1.0, 1.0, 0.0))
    ax.yaxis.set_pane_color((1.0, 1.0, 1.0, 0.0))
    ax.zaxis.set_pane_color((1.0, 1.0, 1.0, 0.0))

    # Eliminar el color de fondo de la figura
    fig.patch.set_alpha(0.0)

    # Mostrar el gráfico
    plt.show()

# Ejecutar la función para generar el gráfico
if __name__ == '__main__':
    generar_grafico_3d_identico()
```
